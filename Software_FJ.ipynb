```python

# Sistema Software FJ
# Primera parte: manejo de clientes

# Importamos herramientas para crear clases abstractas
from abc import ABC, abstractmethod

# Creamos una clase para los errores del sistema
class ErrorSistema(Exception):
    pass


# Clase general para las entidades
class Entidad(ABC):

    def __init__(self, identificacion, nombre):
        self.identificacion = identificacion
        self.nombre = nombre

    @abstractmethod
    def mostrar_informacion(self):
        pass


# Clase Cliente que hereda de Entidad
class Cliente(Entidad):

    def __init__(self, identificacion, nombre, correo, telefono):

        # Validamos que los datos no estén vacíos
        if not identificacion or not nombre:
            raise ErrorSistema("La identificación y el nombre son obligatorios")

        if "@" not in correo:
            raise ErrorSistema("El correo electrónico no es válido")

        if not telefono.isdigit():
            raise ErrorSistema("El teléfono debe contener solo números")

        # Guardamos los datos del cliente
        super().__init__(identificacion, nombre)

        self.__correo = correo
        self.__telefono = telefono

    # Mostramos la información del cliente
    def mostrar_informacion(self):
        print("Identificación:", self.identificacion)
        print("Nombre:", self.nombre)
        print("Correo:", self.__correo)
        print("Teléfono:", self.__telefono)


# Probamos el registro de un cliente
cliente1 = Cliente(
    "12345",
    "Carlos Perez",
    "carlos@gmail.com",
    "3001234567"
)

cliente1.mostrar_informacion()
```

```python
# Segunda parte: clases de servicios

# Creamos una clase general para los servicios
class Servicio(ABC):

    def __init__(self, nombre, precio_base):
        if precio_base <= 0:
            raise ErrorSistema("El precio debe ser mayor que cero")

        self.nombre = nombre
        self.precio_base = precio_base

    @abstractmethod
    def calcular_costo(self, cantidad):
        pass

    @abstractmethod
    def describir(self):
        pass


# Servicio 1: reserva de salas
class ReservaSala(Servicio):

    def calcular_costo(self, horas):
        if horas <= 0:
            raise ErrorSistema("Las horas deben ser mayores que cero")

        return self.precio_base * horas

    def describir(self):
        return "Servicio de reserva de salas"


# Servicio 2: alquiler de equipos
class AlquilerEquipos(Servicio):

    def calcular_costo(self, dias):
        if dias <= 0:
            raise ErrorSistema("Los dias deben ser mayores que cero")

        return self.precio_base * dias

    def describir(self):
        return "Servicio de alquiler de equipos"


# Servicio 3: asesorias especializadas
class Asesoria(Servicio):

    def calcular_costo(self, horas):
        if horas <= 0:
            raise ErrorSistema("Las horas deben ser mayores que cero")

        return self.precio_base * horas

    def describir(self):
        return "Servicio de asesorias especializadas"


# Probamos los tres servicios
sala = ReservaSala("Sala de reuniones", 50000)
equipo = AlquilerEquipos("Computador", 30000)
asesoria = Asesoria("Asesoria de sistemas", 40000)

print(sala.describir())
print("Costo:", sala.calcular_costo(2))

print(equipo.describir())
print("Costo:", equipo.calcular_costo(3))

print(asesoria.describir())
print("Costo:", asesoria.calcular_costo(2))
```

```python
# Tercera parte: sistema de reservas

# Creamos una clase para manejar las reservas
class Reserva:

    def __init__(self, cliente, servicio, duracion):

        if not isinstance(cliente, Cliente):
            raise ErrorSistema("El cliente no es valido")

        if not isinstance(servicio, Servicio):
            raise ErrorSistema("El servicio no es valido")

        if duracion <= 0:
            raise ErrorSistema("La duracion debe ser mayor que cero")

        self.cliente = cliente
        self.servicio = servicio
        self.duracion = duracion
        self.estado = "Pendiente"

    # Metodo para confirmar la reserva
    def confirmar(self):
        if self.estado != "Pendiente":
            raise ErrorSistema("No se puede confirmar esta reserva")

        self.estado = "Confirmada"
        print("La reserva fue confirmada correctamente")

    # Metodo para cancelar la reserva
    def cancelar(self):
        if self.estado == "Cancelada":
            raise ErrorSistema("La reserva ya esta cancelada")

        self.estado = "Cancelada"
        print("La reserva fue cancelada")

    # Metodo para calcular el costo
    def calcular_total(self):
        return self.servicio.calcular_costo(self.duracion)

    # Metodo para mostrar la reserva
    def mostrar_reserva(self):
        print("\n--- INFORMACION DE LA RESERVA ---")
        print("Cliente:", self.cliente.nombre)
        print("Servicio:", self.servicio.nombre)
        print("Duracion:", self.duracion)
        print("Estado:", self.estado)
        print("Costo total:", self.calcular_total())


# Creamos una reserva de prueba
try:
    reserva1 = Reserva(cliente1, sala, 2)
    reserva1.confirmar()
    reserva1.mostrar_reserva()

except ErrorSistema as error:
    print("Error:", error)

finally:
    print("Proceso de reserva finalizado")
```

```python
# Cuarta parte: registro de errores y eventos

import logging

# Configuramos el archivo donde se guardaran los registros
logging.basicConfig(
    filename="logs.txt",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

# Funcion para registrar eventos
def registrar_evento(mensaje):
    logging.info(mensaje)
    print("Registro:", mensaje)


# Funcion para registrar errores
def registrar_error(mensaje):
    logging.error(mensaje)
    print("Error registrado:", mensaje)


# Probamos el registro de una reserva
registrar_evento("El sistema Software FJ fue iniciado")
registrar_evento("Se creo un cliente correctamente")
registrar_evento("Se confirmo una reserva")

# Probamos el registro de un error
try:
    raise ErrorSistema("Se intento realizar una operacion incorrecta")

except ErrorSistema as error:
    registrar_error(str(error))

# Verificamos que el archivo fue creado
with open("logs.txt", "r") as archivo:
    print("\nContenido del archivo de registros:")
    print(archivo.read())
```

```python
class SistemaSoftwareFJ:
    def __init__(self):
        self.clientes = []
        self.servicios = []
        self.reservas = []

    def agregar_cliente(self, cliente):
        self.clientes.append(cliente)
        registrar_evento("Cliente agregado: " + cliente.nombre)

    def agregar_servicio(self, servicio):
        self.servicios.append(servicio)
        registrar_evento("Servicio agregado: " + servicio.nombre)

    def agregar_reserva(self, reserva):
        self.reservas.append(reserva)
        registrar_evento("Reserva agregada para: " + reserva.cliente.nombre)

    def mostrar_clientes(self):
        print("\n--- LISTA DE CLIENTES ---")
        for cliente in self.clientes:
            print(cliente.nombre)

    def mostrar_servicios(self):
        print("\n--- LISTA DE SERVICIOS ---")
        for servicio in self.servicios:
            print(servicio.nombre)

    def mostrar_reservas(self):
        print("\n--- LISTA DE RESERVAS ---")
        for reserva in self.reservas:
            reserva.mostrar_reserva()


sistema = SistemaSoftwareFJ()

sistema.agregar_cliente(cliente1)
sistema.agregar_servicio(sala)
sistema.agregar_servicio(equipo)
sistema.agregar_servicio(asesoria)
sistema.agregar_reserva(reserva1)

sistema.mostrar_clientes()
sistema.mostrar_servicios()
sistema.mostrar_reservas()
```

```python
# Sobrecarga de metodo usando parametros opcionales

class CalculadoraServicios:

    def calcular(self, precio, cantidad=1, descuento=0):
        total = precio * cantidad
        total = total - (total * descuento / 100)
        return total


calculadora = CalculadoraServicios()

print("Costo normal:", calculadora.calcular(50000))
print("Costo por varias horas:", calculadora.calcular(50000, 3))
print("Costo con descuento:", calculadora.calcular(50000, 3, 10))
```

```python
# Encadenamiento de excepciones

def buscar_cliente(identificacion):
    try:
        cliente = next(
            c for c in sistema.clientes
            if c.identificacion == identificacion
        )
        return cliente

    except StopIteration as error:
        raise ErrorSistema(
            "No se encontro el cliente solicitado"
        ) from error


try:
    buscar_cliente("99999")

except ErrorSistema as error:
    print("Error del sistema:", error)
    print("Se produjo un error encadenado")
```
